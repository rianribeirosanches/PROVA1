```python
# 1. Entrada de dados
idade = int(input("Digite a sua idade: "))

# 2. Verificação da idade
if idade < 16:
    print("Entrada permitida apenas com acompanhante")
else:
    # Perguntamos o dinheiro apenas se a pessoa tiver 16 anos ou mais
    dinheiro = float(input("Quanto dinheiro você tem disponível na carteira? R$ "))

    # 3. Verificação do dinheiro
    if dinheiro >= 30.00:
        print("Ingresso comprado! Bom filme")
    else:
        print("Saldo insuficiente para o ingresso")

# 4. Finalização (fora de qualquer estrutura condicional para rodar sempre)
print("Foi bom te atender!")
```
